### Hillslope

```python
import numpy as np

def hillslope_func(timestep, Par, forcing, Fluxes, States):
    """
    Hillslope unit of a semi-distributed FLEX/HBV-style model.

    Parameters
    ----------
    timestep : int
        Current timestep index.
    Par : array-like
        Model parameters in the order:
        [I_max, Ce, Su_max, beta, D, Kf]
    forcing : dict
        Forcing dictionary with keys:
        'Qo'   : observed runoff (not used in state update here)
        'Prec' : precipitation time series
        'Etp'  : potential evapotranspiration time series
    Fluxes : ndarray
        Flux array with columns:
        0 = Ei_dt  : interception evaporation
        1 = Ea_dt  : actual transpiration
        2 = Qf_dt  : fast runoff from fast reservoir
        3 = Qus_dt : preferential percolation / recharge
    States : ndarray
        State array with columns:
        0 = Si : interception storage
        1 = Su : unsaturated storage
        2 = Sf : fast reservoir storage

    Returns
    -------
    Fluxes, States
    """

    # --------------------------------------------------
    # Unpack parameters
    # --------------------------------------------------
    I_max = Par[0]   # interception storage capacity
    Ce    = Par[1]   # evapotranspiration reduction parameter
    Su_max = Par[2]  # maximum unsaturated storage
    beta  = Par[3]   # runoff generation shape parameter
    D     = Par[4]   # split between recharge and fast flow
    Kf    = Par[5]   # fast reservoir recession coefficient

    # --------------------------------------------------
    # Unpack forcing
    # --------------------------------------------------
    Qo   = forcing['Qo']      # not used here, but kept for consistency
    Prec = forcing['Prec']
    Etp  = forcing['Etp']

    # --------------------------------------------------
    # Access state and flux arrays
    # --------------------------------------------------
    Si = States[:, 0]
    Su = States[:, 1]
    Sf = States[:, 2]

    Ei_dt  = Fluxes[:, 0]
    Ea_dt  = Fluxes[:, 1]
    Qf_dt  = Fluxes[:, 2]
    Qus_dt = Fluxes[:, 3]

    # --------------------------------------------------
    # Time handling
    # --------------------------------------------------
    dt = 1.0
    t = timestep

    P_dt  = Prec[t] * dt
    Ep_dt = Etp[t] * dt

    # Previous states
    if t == 0:
        Si_prev = Si[t]
        Su_prev = Su[t]
        Sf_prev = Sf[t]
    else:
        Si_prev = Si[t - 1]
        Su_prev = Su[t - 1]
        Sf_prev = Sf[t - 1]

    # ==================================================
    # 1. INTERCEPTION RESERVOIR
    # ==================================================
    # Throughfall / effective precipitation leaving canopy:
    # Pe = max(0, Si_prev + P - I_max)
    Pe_dt = max(0.0, Si_prev + P_dt - I_max)

    # Temporary interception storage after rainfall input
    Si_temp = min(I_max, Si_prev + P_dt)

    # Interception evaporation
    Ei_dt[t] = min(Ep_dt, Si_temp)

    # Updated interception storage
    Si[t] = Si_temp - Ei_dt[t]

    # Remaining potential evaporation available for transpiration
    Ep_rem = max(0.0, Ep_dt - Ei_dt[t])

    # ==================================================
    # 2. UNSATURATED RESERVOIR
    # ==================================================
    # Wetness-dependent runoff/recharge coefficient
    if Su_max > 0:
        Cr = (Su_prev / Su_max) ** beta
    else:
        Cr = 0.0

    # Limit coefficient to physical range
    Cr = min(max(Cr, 0.0), 1.0)

    # Mobile water generated from throughfall
    Rmobile = Cr * Pe_dt

    # Water retained in the unsaturated zone
    Su_temp = Su_prev + (1.0 - Cr) * Pe_dt

    # Prevent storage exceeding capacity
    if Su_temp > Su_max:
        excess = Su_temp - Su_max
        Su_temp = Su_max
        Rmobile += excess

    # ==================================================
    # 3. TRANSPIRATION
    # ==================================================
    # Soil moisture stress factor
    if Ce * Su_max > 0:
        moisture_factor = min(Su_temp / (Ce * Su_max), 1.0)
    else:
        moisture_factor = 0.0

    Ea_dt[t] = min(Ep_rem * moisture_factor, Su_temp)

    # Updated unsaturated storage after transpiration
    Su[t] = Su_temp - Ea_dt[t]

    # ==================================================
    # 4. PREFERENTIAL PERCOLATION / RECHARGE
    # ==================================================
    Qus_dt[t] = D * Rmobile

    # ==================================================
    # 5. FAST RESERVOIR
    # ==================================================
    # Inflow to fast reservoir is the remaining mobile water
    Rf_dt = (1.0 - D) * Rmobile

    # Update fast storage
    Sf_temp = Sf_prev + Rf_dt

    # Linear fast runoff
    Qf_dt[t] = min(Kf * Sf_temp, Sf_temp)

    # Updated fast storage
    Sf[t] = Sf_temp - Qf_dt[t]

    # --------------------------------------------------
    # Save updated arrays back
    # --------------------------------------------------
    States[:, 0] = Si
    States[:, 1] = Su
    States[:, 2] = Sf

    Fluxes[:, 0] = Ei_dt
    Fluxes[:, 1] = Ea_dt
    Fluxes[:, 2] = Qf_dt
    Fluxes[:, 3] = Qus_dt

    return Fluxes, States
```

### Flextopo

```python
def FLEXtopo(ParPlateau, ParHillslope, ParWetland, ParCatchment, forcing, landscapes):

	# Parameters and constants
	T_lag = ParCatchment[1]
	Ks = ParCatchment[0]
	dt = 1
	t_max = len(forcing['Qo'])

	# Initialize states
	States_plateau = np.zeros((t_max,3))
	States_hillslope = np.zeros((t_max,3))
	States_wetland = np.zeros((t_max,3))
	Ss = np.zeros((t_max,1))

	# Initialize fluxes
	Fluxes_plateau = np.zeros((t_max,4))
	Fluxes_hillslope = np.zeros((t_max,4))
	Fluxes_wetland = np.zeros((t_max,3))
	Qs_dt = np.zeros(t_max)
	Q_tot_dt = np.zeros(t_max)

	# Loop over time
	for t in range(0,t_max):

		# Plateau
		Fluxes_plateau, States_plateau = plateau_func(t, ParPlateau, forcing, Fluxes_plateau, States_plateau)
		# Hillslope
		Fluxes_hillslope, States_hillslope = hillslope_func(t, ParHillslope, forcing, Fluxes_hillslope, States_hillslope)

		# Wetland
		Fluxes_wetland, States_wetland, Ss = wetland_func(t, ParWetland, forcing, Fluxes_wetland, States_wetland, Ss, landscapes[2])

		# Slow Reservoir
		'''UPDATE'''
		Ss[t] = Ss[t] + ... * landscapes[0] + ... * landscapes[1]
		Qs_dt = dt * Ks * Ss[t]
		Ss[t] = Ss[t] - min(Qs_dt,Ss[t])
		if t < t_max - 1:
			Ss[t+1] = Ss[t]

		'''UPDATE'''
		Q_tot_dt[t] = Qs_dt + ... * landscapes[0] + ... * landscapes[1] +  ... * landscapes[2]

	# Offset Q
	weigths = Weigfun(T_lag)

	Qm = np.convolve(Q_tot_dt, weigths)
	Qm = Qm[0:t_max]

	return(Qm)
```
